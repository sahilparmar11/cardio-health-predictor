# Convert Age from days to years

# Identify Outliers

```python
def detect_outliers_iqr(column):
    Q1 = column.quantile(0.25)
    Q3 = column.quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return lower, upper
```

# Visualize Outlier Using Boxplot

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
num_cols = ['age_years', 'height', 'weight', 'ap_hi', 'ap_lo']

plt.figure(figsize=(15,8))
for i, col in enumerate(num_cols, 1):
    plt.subplot(2,3,i)
    sns.boxplot(y=df[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

# Handle Outliers(IQR Capping)

```python
for col in num_cols:
    lower, upper = detect_outliers_iqr(df[col])
    df[col] = np.where(df[col] < lower, lower, df[col])
    df[col] = np.where(df[col] > upper, upper, df[col])
```

```python
sns.boxplot(y=df[col])
plt.title(f"{col} after outlier handling")
plt.show()
```

# Check Blood Pressure Validity

```python
df = df[df['ap_hi'] > df['ap_lo']]
df
```
